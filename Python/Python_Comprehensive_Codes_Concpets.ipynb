````python
# while to reverse a number
num = int(input("Enter a number: "))
reversed_num = 0
original_num = num # keeping it for display purposes
while num > 0: # while digits remain
    digit = num % 10 # get the last digit
    reversed_num = (reversed_num * 10) + digit # shift left and add the digit
    num //= 10

print(f"Original number: {original_num}")
print(f"Reversed number: {reversed_num}")
````

```
num=123, reversed=0
Iteration 1:
  digit = 123 % 10 = 3
  reversed = 0*10 + 3 = 3
  num = 123 // 10 = 12

Iteration 2:  
  digit = 12 % 10 = 2
  reversed = 3*10 + 2 = 32
  num = 12 // 10 = 1

Iteration 3:
  digit = 1 % 10 = 1
  reversed = 32*10 + 1 = 321
  num = 1 // 10 = 0 → Loop ends!

Output: Reversed = 321
```

````python
# Linear Search Algorithm

numbers = [10, 20, 70, 30, 9, 5, 101, 40, 80, 90, 2, 5, 6]
search_value = 70

def linear_search(numbers, search_value):
  for i in range(len(numbers)):
    if numbers[i] == search_value:
      return i

result =  linear_search(numbers, search_value)

if result != -1:
  print(f"Value found at index: {result}")
else:
  print(f"Value not found!!! at any index. Return value {result} ")
````

````python
# Binary Search

def binary_search(numbers, target):
    left, right = 0, len(numbers) - 1

    while left <= right:  # Search space exists
        mid = left + (right - left) // 2  # Avoid overflow

        if numbers[mid] == target:
            return mid
        elif target > numbers[mid]:
            left = mid + 1    # Discard left half
        else:
            right = mid - 1   # Discard right half

    return -1  # Not found

numbers = [1,2,5,6, 10, 14, 16, 19, 20, 24, 70 ,29, 240, 24, 249, 234, 234, 234, 23 ,23 ,2312]
sorted_nums = sorted(numbers)
print(sorted_nums)
result = binary_search(sorted_nums, 2312)
print(f"Found at index {result} in sorted list")
````
